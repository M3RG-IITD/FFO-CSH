```python
import os
from pathlib import Path
from copy import deepcopy

# Base Charge Dictionary
q_values_den = {
    'q_cat': 1.7, 'q_casl': 1.7, 'q_sc2_3': 1.0,
    'q_oc12': -0.5, 'q_oc13': -1.1, 'q_oc22': -1.1,
    'q_ocx': -0.5, 'q_oc14': -0.67, 'q_hoy': 0.42
}

# Delta Corrections for Surface Energy Calculation
delta_se = {
    'q_oc12': +0.02,
    'q_oc13': -0.01,
    'q_ocx':  -0.04,
    'q_oc14': +0.02,
    'q_hoy':  -0.02,
    'q_oc22': -0.04
}

delta_se_14 = {
    'q_oc12': +0.03,
    'q_oc13': -0.01,
    'q_ocx':  -0.04,
    'q_oc14': +0.02,
    'q_hoy':  -0.02,
    'q_oc22': -0.05
}

# Apply delta corrections to base dictionary
def apply_deltas(base, delta):
    updated = deepcopy(base)
    for key, delta_value in delta.items():
        if key in updated:
            updated[key] += delta_value
        else:
            raise KeyError(f"{key} not found in base charge dictionary!")
    return updated

q_values_se = apply_deltas(q_values_den, delta_se)
q_values_se_14 = apply_deltas(q_values_den, delta_se_14)

# Helper Functions
def get_charge(charge_str, q_values):
    try:
        return float(charge_str)
    except ValueError:
        placeholder = charge_str.strip('{}')
        if placeholder in q_values:
            return q_values[placeholder]
        else:
            raise ValueError(f"Unknown placeholder: {charge_str}")

def process_file(file_path, q_values, is_car=True):
    total_charge = 0.0
    with open(file_path, 'r') as file:
        for line in file:
            line = line.strip()
            if not line or line.startswith(('!', 'PBC', '#', '@', '*')):
                continue
            tokens = line.split()
            idx = 8 if is_car else 6
            if len(tokens) > idx:
                charge = get_charge(tokens[idx], q_values)
                total_charge += charge
    file_type = "CAR" if is_car else "MDF"
    print(f"{file_path.name:<40} Total charge ({file_type}): {total_charge: .5f}")

def determine_charge_scheme(folder_name, file_name):
    if "Tob14" in folder_name:
        return q_values_se_14 if any(k in file_name for k in ['Bulk', 'vacuum', 'Cleaved']) else q_values_den
    else:
        return q_values_se if any(k in file_name for k in ['Bulk', 'vacuum', 'Cleaved']) else q_values_den

# Main Loop
base_dir = Path("../../data/structures/templates/template_car_mdf")
folders = ["Tob11_car_mdf", "Tob11H_car_mdf", "Tob14_car_mdf"]

for folder in folders:
    folder_path = base_dir / folder
    car_files = sorted(folder_path.glob("*.car"))
    mdf_files = sorted(folder_path.glob("*.mdf"))

    print(f"\n### Checking files in {folder} ###")

    for car_file in car_files:
        q_values = determine_charge_scheme(folder, car_file.name)
        process_file(car_file, q_values, is_car=True)

    for mdf_file in mdf_files:
        q_values = determine_charge_scheme(folder, mdf_file.name)
        process_file(mdf_file, q_values, is_car=False)
```
