```python
import numpy as np
from scipy.stats import qmc

# Define parameter bounds
param_bounds = {
    "sc_r": (4.30, 4.55),
    "sc_eps": (0.30, 0.51),
    "oc_r": (3.30, 3.80),
    "oc_eps": (0.03, 0.19),
}

# Number of samples
n_samples = 50

# Create LHS sampler
sampler = qmc.LatinHypercube(d=len(param_bounds))
lhs_unit = sampler.random(n=n_samples)

# Scale to actual bounds
lower_bounds = [v[0] for v in param_bounds.values()]
upper_bounds = [v[1] for v in param_bounds.values()]
lhs_scaled = qmc.scale(lhs_unit, lower_bounds, upper_bounds)

# To get a NumPy array
all_params_lhs = lhs_scaled

sc_r_values_n = []
sc_eps_values_n = []
oc_r_values_n = []
oc_eps_values_n = []


# Grid
sc_r_values = np.linspace(4.30, 4.55, 5)
sc_eps_values = np.linspace(0.30, 0.51, 5)

oc_r_values = np.linspace(3.30, 3.80, 5)
oc_eps_values = np.linspace(0.08, 0.19, 5)


for i, (s_r,s_eps) in enumerate(zip(sc_r_values, sc_eps_values)):
    for  j, (o_r,o_eps) in enumerate(zip(oc_r_values, oc_eps_values)):
        # print(f"sc_r: {s_r}, sc_eps: {s_eps}, oc_r: {o_r}, oc_eps: {o_eps}")
        sc_r_values_n.append(s_r)
        sc_eps_values_n.append(s_eps)
        oc_r_values_n.append(o_r)
        oc_eps_values_n.append(o_eps)

all_params_grid = np.array([sc_r_values_n, sc_eps_values_n, oc_r_values_n, oc_eps_values_n]).T
```

```python

all_params = np.vstack((all_params_grid, all_params_lhs))
```

```python
# plot 2*2 plot with sc_r vs sc_eps, oc_r vs oc_eps, sc_r vs oc_r, sc_eps vs oc_eps
import matplotlib.pyplot as plt
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
for i in range(2):
    for j in range(2):
        ax = axs[i, j]
        if i == 0 and j == 0:
            # ax.scatter(all_params_lhs[:, 0], all_params_lhs[:, 1], label='LHS')
            ax.scatter(all_params_grid[:, 0], all_params_grid[:, 1],c = 'r', label='Grid')
            ax.set_title('sc_r vs sc_eps')
            ax.set_xlabel('sc_r')
            ax.set_ylabel('sc_eps')
            # ax.legend()
        elif i == 0 and j == 1:
            # ax.scatter(all_params_lhs[:, 2], all_params_lhs[:, 3], label='LHS')
            ax.scatter(all_params_grid[:, 2], all_params_grid[:, 3],c = 'r', label='Grid')
            ax.set_title('oc_r vs oc_eps')
            ax.set_xlabel('oc_r')
            ax.set_ylabel('oc_eps')
            # ax.legend()
        elif i == 1 and j == 0:
            # ax.scatter(all_params_lhs[:, 0], all_params_lhs[:, 2], label='LHS')
            ax.scatter(all_params_grid[:, 0], all_params_grid[:, 2],c = 'r', label='Grid')
            ax.set_title('sc_r vs oc_r')
            ax.set_xlabel('sc_r')
            ax.set_ylabel('oc_r')
            # ax.legend()
        elif i == 1 and j == 1:
            # ax.scatter(all_params_lhs[:, 1], all_params_lhs[:, 3], label='LHS')
            ax.scatter(all_params_grid[:, 1], all_params_grid[:, 3],c = 'r', label='Grid')
            ax.set_title('sc_eps vs oc_eps')
            ax.set_xlabel('sc_eps')
            ax.set_ylabel('oc_eps')
            ax.legend()
```
