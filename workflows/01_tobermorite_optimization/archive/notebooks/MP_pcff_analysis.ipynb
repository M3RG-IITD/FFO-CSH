```python
import os
import numpy as np
import matplotlib.pyplot as plt
import pickle

# Change to the correct directory
os.chdir('')

# Define paths
LJ_SETS = [f"LJ_set_{i+1}" for i in range(68)]
TOB_STRUCTURES = ["Tob9", "Tob11", "Tob11H", "Tob14"]

source_colors = {"Tob9": "blue", "Tob11": "green", "Tob11H": "red", "Tob14": "purple"}

# Ground truth values (None means missing)
ground_truths = {
    "Tob14": {
        "Density": {"EXP": 2.23},
        "SE004": {"Tariq": 635},
        "BM": {"EXP": 47},
    },
    "Tob11": {
        "Density": {"EXP": 2.46},
        "SE004": {"Tariq": 680},
        "BM": {"EXP": 71},
    },
    "Tob11H": {
        "Density": {"EXP": 2.39},
        "SE004": {"Tariq": 325},
        "BM": {"EXP": 55.35}, # using DFT value
    },
    "Tob9": {
        "Density": {"EXP": 2.865},
        "SE004": {"Tariq": 805},
        "BM": {"EXP": 115},
    },
}

# Colors for each GT source
source_linestyles = {"DFT": "--", "EXP": "-.", "Tariq": ":"}
```

```python

sc_r_values_n = []
sc_eps_values_n = []
oc_r_values_n = []
oc_eps_values_n = []

# Grid 1
sc_r_values = np.linspace(4.5, 4.8, 5)
sc_eps_values = np.linspace(0.22, 0.40, 5)

oc_r_values = np.linspace(3.45, 3.85, 5)
oc_eps_values = np.linspace(0.06, 0.16, 5)


for i, (s_r,s_eps) in enumerate(zip(sc_r_values, sc_eps_values)):
    for  j, (o_r,o_eps) in enumerate(zip(oc_r_values, oc_eps_values)):
        # print(f"sc_r: {s_r}, sc_eps: {s_eps}, oc_r: {o_r}, oc_eps: {o_eps}")
        sc_r_values_n.append(s_r)
        sc_eps_values_n.append(s_eps)
        oc_r_values_n.append(o_r)
        oc_eps_values_n.append(o_eps)


# Grid 2
sc_r_values = np.linspace(4.2, 4.5, 5)
sc_eps_values = np.linspace(0.04, 0.22, 5)

oc_r_values = np.linspace(3.05, 3.45, 5)
oc_eps_values = np.linspace(0.0001, 0.06, 5)


for i, (s_r,s_eps) in enumerate(zip(sc_r_values, sc_eps_values)):
    for  j, (o_r,o_eps) in enumerate(zip(oc_r_values, oc_eps_values)):
        # print(f"sc_r: {s_r}, sc_eps: {s_eps}, oc_r: {o_r}, oc_eps: {o_eps}")
        sc_r_values_n.append(s_r)
        sc_eps_values_n.append(s_eps)
        oc_r_values_n.append(o_r)
        oc_eps_values_n.append(o_eps)


# Grid 3
sc_r_values = np.linspace(4.2, 4.5, 4)[:-1]
sc_eps_values = np.linspace(0.04, 0.22, 4)[:-1]

oc_r_values = np.linspace(3.45, 3.85, 4)[1:]
oc_eps_values = np.linspace(0.06, 0.16, 4)[1:]

for i, (s_r,s_eps) in enumerate(zip(sc_r_values, sc_eps_values)):
    for  j, (o_r,o_eps) in enumerate(zip(oc_r_values, oc_eps_values)):
        # print(f"sc_r: {s_r}, sc_eps: {s_eps}, oc_r: {o_r}, oc_eps: {o_eps}")
        sc_r_values_n.append(s_r)
        sc_eps_values_n.append(s_eps)
        oc_r_values_n.append(o_r)
        oc_eps_values_n.append(o_eps)
            

# Grid 4
sc_r_values = np.linspace(4.5, 4.8, 4)[1:]
sc_eps_values = np.linspace(0.22, 0.40, 4)[1:]

oc_r_values = np.linspace(3.05, 3.45, 4)[:-1]
oc_eps_values = np.linspace(0.0001, 0.06, 4)[:-1]

for i, (s_r,s_eps) in enumerate(zip(sc_r_values, sc_eps_values)):
    for  j, (o_r,o_eps) in enumerate(zip(oc_r_values, oc_eps_values)):
        # print(f"sc_r: {s_r}, sc_eps: {s_eps}, oc_r: {o_r}, oc_eps: {o_eps}")
        sc_r_values_n.append(s_r)
        sc_eps_values_n.append(s_eps)
        oc_r_values_n.append(o_r)
        oc_eps_values_n.append(o_eps)
        
        
# plot sc_r_values_n vs oc_r_values_n
plt.figure(figsize=(10, 6))
plt.scatter(sc_r_values_n[:50], oc_r_values_n[:50], c='orange', label='run1')
plt.scatter(sc_r_values_n[50:], oc_r_values_n[50:], c='blue', label='run2')
plt.xlabel('sc_r')
plt.ylabel('oc_r')
plt.title('sc_r vs oc_r grid')
plt.grid()
```

```python
#load den_results 50
with open("results_den50.pkl", "rb") as f:
    den_results_50 = pickle.load(f)


#load den_results last 18
with open("results_den68.pkl", "rb") as f:
    den_results_18 = pickle.load(f)



# Start with a copy to avoid modifying the original
den_results = {'density': den_results_50['density'].copy()}

# Now update/append from den_results_18
for key, value in den_results_18['density'].items():
    if key in den_results['density']:
        den_results['density'][key].extend(value)
    else:
        den_results['density'][key] = value
```

```python
plt.figure(figsize=(25, 6))
ax = plt.gca()

for tob_structure in TOB_STRUCTURES[1:]:
    data = np.array(den_results['density'][tob_structure])
    if data.size == 0:
        continue
    ax.plot(range(1, len(LJ_SETS) + 1), data[:, 0], label=tob_structure, marker="o", color=source_colors[tob_structure])

    # GT lines for Density
    gt_sources = ground_truths[tob_structure].get("Density", {})
    for source, value in gt_sources.items():
        if value is not None:
            ax.hlines(value, 1, len(LJ_SETS),
                      colors=source_colors[tob_structure],
                      linestyles=source_linestyles[source],
                      label=f"{source} GT ({tob_structure})")

ax.set_xlabel("LJ Set Number")
ax.set_ylabel("Density")
ax.set_title(f"LJ Set vs Density")
ax.set_xticks(range(1, len(LJ_SETS)+1))
ax.set_xticklabels([f"S{i+1}" for i in range(len(LJ_SETS))])
ax.grid()
handles, labels = ax.get_legend_handles_labels()
unique = dict(zip(labels, handles))
ax.legend(unique.values(), unique.keys(), loc='center left', bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.show()
```

```python

for tob_structure in TOB_STRUCTURES:
    den = [i[0] for i in (den_results['density'][tob_structure])]

    plt.scatter(sc_r_values_n, oc_r_values_n, c=den, cmap='viridis', vmin=min(den), vmax=max(den))
    plt.colorbar(label='Density')
    plt.xlabel('sc_r')
    plt.ylabel('oc_r')
    den_gt = ground_truths[tob_structure]["Density"]['EXP']
    plt.title(f'{tob_structure} (EXP Density = {den_gt})')
    plt.legend()
    plt.grid()
    plt.show()

```

```python
#load se_results 50
with open("results_se50.pkl", "rb") as f:
    se_results_50 = pickle.load(f)


#load se_results last 18
with open("results_se68.pkl", "rb") as f:
    se_results_18 = pickle.load(f)


# Start with a copy to avoid modifying the original
se_results = se_results_50.copy()

# Now update/append from se_results_18
for key, value in se_results_18.items():
    if key in se_results:
        se_results[key].extend(value)
    else:
        se_results[key] = value
```

```python
plt.figure(figsize=(25, 6))
ax = plt.gca()

for tob_structure in TOB_STRUCTURES[1:]:
    SE_vals = [v[0] for v in se_results[tob_structure]]
    errors = [v[1] for v in se_results[tob_structure]]
    # ax.errorbar(range(1, 6), SE_vals, yerr=errors, label=tob_structure, marker='o', capsize=5)
    ax.plot(range(1, len(LJ_SETS)+1), SE_vals, marker='o', label=tob_structure, color=source_colors[tob_structure])

    # GT lines for SE004
    gt_sources = ground_truths[tob_structure].get("SE004", {})
    for source, value in gt_sources.items():
        if value is not None:
            ax.hlines(value, 1, len(LJ_SETS),
                      colors=source_colors[tob_structure],
                      linestyles=source_linestyles[source],
                      label=f"{source} GT ({tob_structure})")

ax.set_title("LJ Parameter Set vs Surface Energy")
ax.set_xlabel("LJ Set Number")
ax.set_ylabel("Surface Energy (mJ/m²)")
ax.set_xticks(range(1, len(LJ_SETS)+1))
ax.set_xticklabels([f"S{i+1}" for i in range(len(LJ_SETS))])
ax.grid(True)
handles, labels = ax.get_legend_handles_labels()
unique = dict(zip(labels, handles))
ax.legend(unique.values(), unique.keys(), loc='center left', bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.show()
```

```python
for tob_structure in TOB_STRUCTURES:
    SE_vals = [v[0] for v in se_results[tob_structure]]
    SE_vals_n = []
    for i in SE_vals:
        if i is None:
            SE_vals_n.append(0)
        elif i<=0.0:
            SE_vals_n.append(0)
        else:
            SE_vals_n.append(i)
    
    SE_vals_n = np.array(SE_vals_n)
    SE_vals_n1 = SE_vals_n.copy()
    SE_vals_n1 = np.sort(SE_vals_n1)
    min2 = np.unique(SE_vals_n1)[1]

    # put red circle if SE_vals_n is 0
    for i, val in enumerate(SE_vals_n):
        if val == 0:
            plt.scatter(sc_r_values_n[i], oc_r_values_n[i], c='red', s=100, edgecolors='black')
    
    plt.scatter(sc_r_values_n, oc_r_values_n, c=SE_vals_n, cmap='viridis', vmin=min2, vmax=max(SE_vals_n))
    
    plt.colorbar(label='Surface Energy')
    plt.xlabel('sc_r')
    plt.ylabel('oc_r')
    se_gt = ground_truths[tob_structure]["SE004"]['Tariq']
    plt.title(f'{tob_structure} (Tariq SE = {se_gt})')
    plt.legend()
    plt.grid()
    plt.show()
```
